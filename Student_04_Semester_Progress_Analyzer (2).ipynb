## 3. Explore the data

```python
# TODO: Display the shape and column names
print("dataset shape ", df.shape)
print("column name:" , df.columns)

# TODO: Inspect data types and missing values
df.info()
print("missing value in each columns")
print(df.isnull().sum())
# TODO: Count duplicate rows
print("number of duplicate rows:",df.duplicated().sum())


```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
 

# TODO: Calculate average, maximum and minimum

sections = ["Semester_1", "Semester_2", "Semester_3"]
marks = df[sections].to_numpy()

print("Overall average:", np.mean(marks))
print("Highest mark:", np.max(marks))
print("Lowest mark:", np.min(marks))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)

# Calculate average marks of each student
df["average"] = df[["Semester_1", "Semester_2", "Semester_3"]].mean(axis=1)

# Calculate progress from Semester 1 to Semester 3
df["progress"] = df["Semester_3"] - df["Semester_1"]


# TODO: Filter records using a condition

# Select students whose marks have improved
improved_students = df[df["progress"] > 0]

print("Improved students:")
display(improved_students)


# TODO: Sort records and display top/bottom results

# Sort students by average marks
sorted_students = df.sort_values(by="average", ascending=False)

# Display top five students
print("Top five students:")
display(sorted_students.head())

# Display bottom five students
print("Bottom five students:")
display(sorted_students.tail())
```
